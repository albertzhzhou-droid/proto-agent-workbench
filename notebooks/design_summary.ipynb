# Proto Design Summary

Local notebook fixture for proto-agent's lightweight notebook runner.

```python
import json
import os
from pathlib import Path

from proto_agent.compiler import compile_design
from proto_agent.sequence import validate_sequences

design_path = 'designs/toggle_switch.proto'
run_dir = Path(os.environ['PROTO_AGENT_RUN_DIR'])
ir, diagnostics = compile_design(design_path)
sequence_report, sequence_diagnostics = validate_sequences(design_path)
payload = {
    'ok': ir is not None and sequence_report['ok'],
    'design_id': ir['design_id'] if ir else None,
    'construct_count': len(ir.get('constructs', [])) if ir else 0,
    'sequence_ok': sequence_report['ok'],
}
(run_dir / 'notebook_summary.json').write_text(json.dumps(payload, indent=2) + '\n', encoding='utf-8')
print(json.dumps(payload, indent=2))
```
